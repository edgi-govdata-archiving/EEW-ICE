
Load DDP data for detention facilities daily populations

```python
url = "https://github.com/deportationdata/ice/raw/00de92b77d8074f38df212f18ef9b4c39fada286/data/facilities-daily-population-latest.parquet" #this file is no longer updated from DDP and fields would need to be calculated
#from "Detention facility stints" going forward: https://ice-detention-stints.apps.deportationdata.org/
#Download > Filter to desired date range > calculate daily n of each demographic category above by facility > associate maximums with other facility data
#There are also other demographic fields there that could be calculated and analyzed
fac_pop = pd.read_parquet(url, filesystem=HTTPFileSystem())
fac_pop
```

```python
fac_pop[fac_pop["detention_facility_code"].isna()]
```

```python
#Fix missing "DETLOC" detention facility codes for detention_fac_loc -
#could not find federal codes specific to these facilities,
#so assigned unique IDs based on facility name and DDP Director Graeme Blair suggestions

detention_fac.loc[
    detention_fac["name"] == "Wichita County Jail",
    "detention_facility_code"
] = "WICHTTX" #This is the code recommended by Graeme Blair - codirector of DDP

detention_fac.loc[
    detention_fac["name"] == "Midwest Regional Reception Center",
    "detention_facility_code"
] = "MIDWRRC" #This is a unique placeholder code until we get a code for this facility. Graeme suggests a code is likely to be reported in the next detention dataset from ICE
```

```python
print(fac_pop["date"].dtype)
```

```python
fac_pop["date"] = pd.to_datetime(fac_pop["date"]).dt.date
```

```python
populations_2025 = fac_pop[fac_pop["date"] >= pd.to_datetime("2025-01-20").date()]
```

```python
maximum_daily_populations_2025 = populations_2025.groupby("detention_facility_code").max()[["n_detained", "n_detained_at_midnight", "n_detained_male", "n_detained_female", "n_detained_convicted_criminal", "n_detained_possibly_under_18", "detention_facility"]]
maximum_daily_populations_2025
```

```python
detention_fac = detention_fac.merge(maximum_daily_populations_2025, on="detention_facility_code", how="left")
detention_fac
```

```python
facs = detention_fac[["detention_facility_code", "name", "latitude", "longitude"]]
```

```python
facs.rename(columns={"latitude": "FAC_LAT", "longitude": "FAC_LONG"}, inplace=True)
facs
```

#Clustering ICE detention facilities by EJScreen Indicators




```python
ice_fac_ejscreen = pandas.read_csv("/content/drive/Shareddrives/EDGI - Shared NEW/12_Environmental_Enforcement_Watch/01_Project_Management/ICE Hazards/detention_fac_ejscreen.csv")
```

```python
columns = ice_fac_ejscreen.columns.tolist()
columns
```

Check if any facilities have NaN for any air indicator

```python
indicators = ['no2',
              'dpm',
              'pm',
              #'rateasthma',
              #'ratecancer',
              'proximity.npl',
              'proximity.rmp',
              'proximity.tsdf',
              'proximity.npdes',
              'ust',
              'n_detained_female',
              'n_detained_possibly_under_18']
```

```python
for col in indicators:
    ice_fac_ejscreen[col] = pandas.to_numeric(
        ice_fac_ejscreen[col],
        errors='coerce'
    )
```

```python
for i in indicators:
  print(i, ice_fac_ejscreen.loc[ice_fac_ejscreen[i].isna()].shape[0])
```

```python
ice_fac_ejscreen.set_index("name", inplace=True)
```

```python
ice_fac_ejscreen = (
    ice_fac_ejscreen
    .dropna(subset=indicators)
    [indicators]
)
ice_fac_ejscreen
```

```python
from sklearn.cluster import KMeans

#ejscreen_norm=(ejscreen-ejscreen.min())/(ejscreen.max()-ejscreen.min()) # min/max normalization
ice_fac_ejscreen_norm=(ice_fac_ejscreen-ice_fac_ejscreen.mean())/ice_fac_ejscreen.std() # mean normalization

clustering_kmeans = KMeans(n_clusters=4, random_state = 0, n_init='auto')
ice_fac_ejscreen_norm['clusters'] = clustering_kmeans.fit_predict(ice_fac_ejscreen_norm)
ice_fac_ejscreen_norm
```

```python
from sklearn.decomposition import PCA
import pandas as pd
import plotly.express as px

pca_num_components = 2

# PCA
X = ice_fac_ejscreen_norm.drop(columns=['clusters'])

reduced_data = PCA(n_components=pca_num_components).fit_transform(X)

results = pd.DataFrame(reduced_data, columns=['pca1', 'pca2'])

# add labels + clusters
results['cluster'] = ice_fac_ejscreen_norm['clusters'].values.astype(str)
results['name'] = ice_fac_ejscreen_norm.index

#plot
fig = px.scatter(
    results,
    x='pca1',
    y='pca2',
    color='cluster',
    hover_name='name',
    category_orders={'cluster': ['0', '1', '2', '3']},
    color_discrete_sequence=px.colors.qualitative.Set1
)
fig.update_xaxes(range=[results.pca1.min()*1.05, results.pca1.max()*1.05])
fig.update_yaxes(scaleanchor="x", scaleratio=1)

fig.show()
```
